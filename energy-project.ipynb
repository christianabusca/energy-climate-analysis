```python
# FINDING
sea_countries = ['Philippines', 'Indonesia', 'Vietnam', 
                 'Thailand', 'Myanmar', 'East Timor', 
                 'Malaysia', 'Singapore', 'Cambodia',
                'Laos', 'Brunei']
sea_df = df[df['country'].isin(sea_countries)]
sea_df
```

```python
null_sea = sea_df.isnull().groupby(sea_df['country']).sum()
null_sea
```

```python
rows_sea = sea_df[sea_df['year'] >= 2004].groupby('country')['year'].size()
rows_sea

```

```python
sea_df = sea_df[sea_df['year'] >= 2005]
sea_df.shape
```

```python
dict(sea_df.isnull().sum())
```

```python
null_sea = sea_df.isnull().groupby(sea_df['country']).sum()
null_sea
```

```python
selected_columns = [
    'country', 'year', 'iso_code',
    'population', 'gdp',
    'fossil_share_energy',
    'renewables_share_energy',
    'solar_share_energy',
    'wind_share_energy',
    'hydro_share_energy',
    'low_carbon_share_energy',
    'primary_energy_consumption',
    'fossil_fuel_consumption',
    'renewables_consumption',
    'energy_per_capita',
    'renewables_cons_change_pct',
    'fossil_cons_change_pct',
    'greenhouse_gas_emissions',
    'carbon_intensity_elec',
    'energy_per_gdp'
]

f_sea_df = sea_df[selected_columns].copy()
f_sea_df.shape
```

```python
nulls = f_sea_df.isnull().groupby(sea_df['country']).sum()
nulls
```

```python
# FINDING
top_sea = ['Philippines', 'Indonesia', 'Vietnam', 
                 'Thailand', 
                 'Malaysia', 'Singapore']
sea_six_df = f_sea_df[f_sea_df['country'].isin(top_sea)]
sea_six_df
```

```python
sea_six_df.isnull().groupby(sea_df['country']).sum()
```

```python
# Year-over-Year (YoY) change

sea_six_df['yoy_rse'] = sea_six_df.sort_values('year').groupby('country')['renewables_share_energy'].pct_change() * 100
sea_six_df['yoy_fse'] = sea_six_df.sort_values('year').groupby('country')['fossil_share_energy'].pct_change() * 100
```

```python
# Cumulative Change Since Baseline
sea_six_df = sea_six_df.sort_values('year')

sea_six_df['cum_rse'] = sea_six_df.groupby('country')['renewables_share_energy'].transform(lambda x: (x - x.iloc[0]) / x.iloc[0] * 100)
sea_six_df['cum_fse'] = sea_six_df.groupby('country')['fossil_share_energy'].transform(lambda x: (x - x.iloc[0]) / x.iloc[0] * 100)
```
