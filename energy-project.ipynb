```python
# Year-over-Year (YoY) change

sea_six_df['yoy_rse'] = sea_six_df.sort_values('year').groupby('country')['renewables_share_energy'].pct_change() * 100
sea_six_df['yoy_fse'] = sea_six_df.sort_values('year').groupby('country')['fossil_share_energy'].pct_change() * 100
```

```python
# Cumulative Change Since Baseline
sea_six_df = sea_six_df.sort_values('year')

sea_six_df['cum_rse'] = sea_six_df.groupby('country')['renewables_share_energy'].transform(lambda x: (x - x.iloc[0]) / x.iloc[0] * 100)
sea_six_df['cum_fse'] = sea_six_df.groupby('country')['fossil_share_energy'].transform(lambda x: (x - x.iloc[0]) / x.iloc[0] * 100)
```

# CURRENT ANALYZATION

Based on the 2024 data, **Singapore** is undergoing the fastest renewable energy transition, recording a Year-over-Year (YoY) increase in renewable energy share of +10.55 percentage points this 2024 from -5.37% last 2023. This substantial rise indicates a rapid acceleration in the integration of renewable sources into its national energy mix within a single year.

In contrast, **Indonesia** leads in reducing fossil fuel dependence, posting the largest YoY decline in fossil energy share at −0.29 percentage points this 2024, from -0.54% last 2023, and +0.27% last 2006. Although the value is negative, it signifies the most aggressive reduction among the Southeast Asian countries in the dataset. This trend suggests that Indonesia is making the fastest progress in cleaning its energy grid relative to its regional peers.

Furthermore, the country that has transformed the most since 2005 is Malaysia, with a cumulative increase of 284% in their renewable energy share. This indicates a long-term structural shift in their national energy strategy, moving from a [Low/Medium] baseline to becoming a regional leader in adoption over the past two decades.

While the cumulative renewable share energy shows growth, cumumulative fossil share energy reveals the true retreat of traditional power. Vietnam is the leader in fossil fuel phase-out, having reduced their fossil share by a cumulative -10.67% since 2005. This demonstrates that their renewable growth is actively displacing carbon-intensive sources rather than just supplementing them.

Shockingly, the data for Philippines reveals a 'Carbon Regression' since 2005, characterized by a negative cumulative renewable share energy and a positive cumulative fossil share energy. This indicates that the country’s energy transition has not only stalled but reversed, with fossil fuels expanding their dominance in the energy mix. This likely stems from Economic Growth and Energy Security, positioning them as a primary outlier in the regional transition.

Moreover, in terms of their energy gap, where you will subtract the Fossil Share Energy to the Renewable Share energy, based on the calculated data, Vietnam is still the leader where it succesfully eradicated 19% of fossil fuel use in the country from 2005 up until 2024. Furthermore, Singapore on the otherhand remains the most fossil-dependent nation within the group. Its energy gap moved from 99.450% in 2005 to 98.910% in 2024, closing by only 0.54 percentage points. Its fossil fuel share barely budged, decreasing by just 0.27% over nearly two decades. 

Unlike its neighbors, the Philippines has seen its energy gap widen. In 2005, it had the smallest gap in the region at 66.694. By 2024, that gap grew to 74.364, an increase of 7.67 percentage points. This is because its fossil fuel share rose from 83.347% to 87.182% (+3.835%) while its renewable share actually dropped. With the provided data on the Philippines in relation to fossil and renewable share energy. This positions Philippines as the only country in the group moving backwards on energy transition, raising concerns about policy direction.

```python
import plotly.graph_objects as go
import pandas as pd

# Filter for 2024 data only
df_2024 = sea_six_df[sea_six_df['year'] == 2024].copy()

# Get 2005 baseline values for hover
df_2005 = sea_six_df[sea_six_df['year'] == 2005][['country', 'renewables_share_energy']].rename(
    columns={'renewables_share_energy': 'baseline_2005'}
)
df_2024 = df_2024.merge(df_2005, on='country', how='left')

# Sort by cumulative change (DESCENDING for highest at top)
df_sorted_by_value = df_2024.sort_values('cum_rse', ascending=False)
df_sorted_alphabetically = df_2024.sort_values('country', ascending=True)

# Color scheme
def get_colors(df):
    colors = []
    for idx, row in df.iterrows():
        if row['cum_rse'] < 0:
            colors.append('#E74C3C')  # Red for negative
        elif row['cum_rse'] > 200:
            colors.append('#27AE60')  # Dark green for >200%
        elif row['cum_rse'] > 150:
            colors.append('#2ECC71')  # Medium green for 150-200%
        else:
            colors.append('#58D68D')  # Light green for <150%
    return colors

# Create figure
fig = go.Figure()

# Trace 1: Sorted by value (visible by default)
fig.add_trace(go.Bar(
    y=df_sorted_by_value['country'],
    x=df_sorted_by_value['cum_rse'],
    orientation='h',
    marker=dict(
        color=get_colors(df_sorted_by_value),
        line=dict(color='#2C3E50', width=1.5)
    ),
    # FIXED: Round to 2 decimals, better formatting
    text=[f"+{val:.2f}%" if val > 0 else f"{val:.2f}%" for val in df_sorted_by_value['cum_rse']],
    textposition='outside',
    textfont=dict(size=14, color='#2C3E50', family='Arial', weight='bold'),
    hovertemplate="<b>%{y}</b><br>" +
                  "<b>Cumulative Change:</b> %{x:.2f}%<br>" +
                  "2005 Baseline: %{customdata[0]:.2f}%<br>" +
                  "2024 Current: %{customdata[1]:.2f}%<br>" +
                  "<extra></extra>",
    customdata=df_sorted_by_value[['baseline_2005', 'renewables_share_energy']].values,
    visible=True
))

# Trace 2: Sorted alphabetically (hidden by default)
fig.add_trace(go.Bar(
    y=df_sorted_alphabetically['country'],
    x=df_sorted_alphabetically['cum_rse'],
    orientation='h',
    marker=dict(
        color=get_colors(df_sorted_alphabetically),
        line=dict(color='#2C3E50', width=1.5)
    ),
    text=[f"+{val:.2f}%" if val > 0 else f"{val:.2f}%" for val in df_sorted_alphabetically['cum_rse']],
    textposition='outside',
    textfont=dict(size=14, color='#2C3E50', family='Arial', weight='bold'),
    hovertemplate="<b>%{y}</b><br>" +
                  "<b>Cumulative Change:</b> %{x:.2f}%<br>" +
                  "2005 Baseline: %{customdata[0]:.2f}%<br>" +
                  "2024 Current: %{customdata[1]:.2f}%<br>" +
                  "<extra></extra>",
    customdata=df_sorted_alphabetically[['baseline_2005', 'renewables_share_energy']].values,
    visible=False
))

# Improved dropdown - smaller, cleaner
fig.update_layout(
    updatemenus=[dict(
        buttons=[
            dict(
                label="Highest First",
                method="update",
                args=[{"visible": [True, False]}]
            ),
            dict(
                label="Alphabetical",
                method="update",
                args=[{"visible": [False, True]}]
            )
        ],
        x=0.01,
        y=1.05,
        xanchor='left',
        yanchor='top',
        bgcolor="#FFFFFF",
        bordercolor="#95A5A6",
        borderwidth=1,
        font=dict(size=11, color='#2C3E50')
    )]
)

# Layout - NO SUBTITLE, cleaner spacing
fig.update_layout(
    title={
        'text': "Energy Transformation Champions: Cumulative Change Since 2005",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 22, 'color': '#2C3E50', 'family': 'Arial Black'}
    },
    
    xaxis=dict(
        title="Cumulative % Change",
        title_font_size=14,
        title_font_color='#2C3E50',
        showgrid=True,
        gridcolor='#E8E8E8',
        gridwidth=1,
        zeroline=True,
        zerolinecolor='#34495E',
        zerolinewidth=2.5,
        showline=True,
        linewidth=2,
        linecolor='#2C3E50',
        tickfont=dict(size=12),
        dtick=50
    ),
    
    yaxis=dict(
        title="",
        showgrid=False,
        showline=False,
        tickfont=dict(size=14, color='#2C3E50', family='Arial')
    ),
    
    plot_bgcolor='white',
    paper_bgcolor='#F8F9FA',
    
    # INCREASED height for better spacing
    height=550,
    width=1100,
    # INCREASED margins - especially right for labels
    margin=dict(l=120, r=200, t=100, b=80),
    
    hovermode='closest',
    showlegend=False
)

# Thicker zero line
fig.add_vline(x=0, line_color='#34495E', line_width=2.5, opacity=0.8)

fig.show()
```
